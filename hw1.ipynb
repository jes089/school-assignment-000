```python
import requests
import bz2
"""
Fetches a BZ2-compressed file from a URL and returns a file-like object
that can be streamed line by line.

Args:
    url (str): The URL of the BZ2-compressed file.

Returns:
    bz2.BZ2File: A file-like object for reading decompressed content.
"""
def open_bz2_from_url(url):
    try:
        response = requests.get(url, stream=True)
        response.raise_for_status()

        # Wrap the raw response in a BZ2File for decompression
        return bz2.BZ2File(response.raw, mode='rb')

    except requests.exceptions.RequestException as e:
        print(f"Error fetching URL: {e}")
        return None
    except OSError as e:
        print(f"Error opening BZ2 data: {e}")
        return None
```
